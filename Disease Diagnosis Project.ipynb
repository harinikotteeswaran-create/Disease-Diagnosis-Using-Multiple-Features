##Data Cleaning

```python
df.rename(columns={'Hx Smoking': 'Smoking History',
                   'Hx Radiothreapy': 'Radiotherapy History',
                   'Pathology': 'Types of Thyroid Cancer (Pathology)',
                   'T': 'Tumor',
                   'N': 'Lymph Nodes',
                   'M': 'Cancer Metastasis',
                  'Response' : 'Treatment Response'}, inplace=True)
df
```

```python
df['Tumor'].replace({'T1a' : 'microscopic' , 'T1b' : 'small' ,
                     'T2' : 'intermediate' ,
                     'T3a' : 'local' ,'T3b' : 'invasive' ,
                     'T4a' : 'extended' ,
                     'T4b' : 'advanced'} , inplace =True)
df
```

```python
df['Lymph Nodes'].replace({'N0': 'contained','N1a': 'spread', 'N1b': 'spread'} , inplace =True)
df
```

```python
df['Cancer Metastasis'].replace({'M0': 'absent' , 'M1' : 'present'} ,inplace =True )
df
```

```python
df['Treatment Response'].replace({
    'Structural Incomplete': 'Incomplete',
    'Biochemical Incomplete': 'Incomplete'
}, inplace=True)
df
```

```python
df.drop_duplicates(inplace=True)
df
```
